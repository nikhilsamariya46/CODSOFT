```python
num1 = int (input("What is the first integer"))
num2 = int (input("What is the second integer"))
operator_sign = input("the name of operator sign there is only four operator sign avaliable +,-,/,* ")
if operator_sign == "+":
    print(num1+num2)
elif operator_sign == "-":
    print(num2-num1)
elif operator_sign == "*":
    print (num1*num2)
elif operator_sign == "/":
    if num2 != 0:
        print(num1/num2)
    else:
        print("CANNOT DIVIDE BY ZERO")
else:
   print("INVALID OPERATOR")
```
