```python
task= []
task_range = int(input("HOW MANY TASK DO YOU WANT TO ADD"))
for i  in range(task_range):
    task.append(input("YOUR TASK"))
USER_INPUT = input("IF YOU WANT TO ADD OR REMOVE PRINT YES OTHERWISE NO")
if USER_INPUT== "YES":
    ADD = input("IF YOU WANT TO ADD PRINT YES OTHERWISE NO")
    if ADD =="YES":
        Range = int(input("HOW MANY TASK DO YOU WANT TO ADD"))
        for g in range(Range):
            task.append(input("TASK"))
        print(task)
    elif ADD == "NO":
        Remove_range = int(input("HOW MANY TASK DO YOU WANT TO REMOVE: "))
        for j in range(Remove_range):
            task_name = input("TASK NAME: ")
            if task_name in task:
                task.remove(task_name)
            else:
                print("TASK NOT FOUND")
        print(task)
else:
    print(task)
```
